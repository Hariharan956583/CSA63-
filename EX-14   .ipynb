```python
import hashlib
from google.colab import files

print("="*60)
print("      MALWARE DETECTION USING SHA-256 HASH")
print("="*60)

malware_database = {
    "2cf24dba5fb0a30e26e83b2ac5b9e29e1b161e5c1fa7425e73043362938b9824": "Virus",
    "5d41402abc4b2a76b9719d911017c592e4b8f8b0d7f8d5a8c4b3d9e1f7e8c1aa": "Trojan",
    "9e107d9d372bb6826bd81d3542a419d6e0b1b5a7b2f8b9c3d4e5f6a7b8c9d012": "Ransomware"
}

def calculate_sha256(filename):
    sha256 = hashlib.sha256()

    with open(filename, "rb") as file:
        while True:
            data = file.read(4096)
            if not data:
                break
            sha256.update(data)

    return sha256.hexdigest()

print("\nUpload a file to scan...\n")

uploaded = files.upload()

filename = list(uploaded.keys())[0]

print("\nFile Uploaded Successfully!")
print("File Name :", filename)

file_hash = calculate_sha256(filename)

print("\nCalculated SHA-256 Hash:")
print(file_hash)

print("\nScanning File...")

if file_hash in malware_database:
    print("\n############################################")
    print("⚠ MALWARE DETECTED!")
    print("Malware Type :", malware_database[file_hash])
    print("Status       : Dangerous File")
    print("############################################")

else:
    print("\n############################################")
    print("✓ FILE IS SAFE")
    print("Status       : Hash Not Found")
    print("############################################")

print("\nDatabase contains", len(malware_database), "known malware hashes.")

choice = input("\nDo you want to add a new malware hash? (yes/no): ").lower()

if choice == "yes":
    new_hash = input("Enter SHA-256 Hash : ")
    malware_name = input("Enter Malware Name : ")

    malware_database[new_hash] = malware_name

    print("\nNew malware added successfully!")

    print("\nUpdated Malware Database:")
    for h, m in malware_database.items():
        print(m, ":", h)

else:
    print("\nDatabase not modified.")

print("\nProgram Completed Successfully.")
```
