```python
 SEGMENTATION_POLICY = {

    ("Guest", "Guest"): True,
    ("Guest", "Corporate"): False,
    ("Guest", "Medical"): False,

    ("Corporate", "Corporate"): True,
    ("Corporate", "Medical"): False,

    ("Admin", "Medical"): True,
    ("Admin", "Corporate"): True,
    ("Admin", "Admin"): True,
}

 def can_communicate(src_segment, dst_segment, policy=SEGMENTATION_POLICY):
    """
    Determines whether communication between
    two network segments is allowed.

    Any undefined segment pair is denied
    (Fail-safe Default).
    """
    return policy.get((src_segment, dst_segment), False)
 def test_experiment6():

    result1 = can_communicate("Guest", "Medical")
    print("Guest -> Medical:", result1)

    assert result1 is False
    result2 = can_communicate("Guest", "Corporate")
    print("Guest -> Corporate:", result2)
    assert result2 is False
    result3 = can_communicate("Admin", "Medical")
    print("Admin -> Medical:", result3)
    assert result3 is True
    result4 = can_communicate("Admin", "Corporate")
    print("Admin -> Corporate:", result4)
    assert result4 is True
    result5 = can_communicate("Corporate", "Corporate")
    print("Corporate -> Corporate:", result5)
    assert result5 is True
    result6 = can_communicate("IoT", "Corporate")
    print("IoT -> Corporate:", result6)
    assert result6 is False
    print("\nAll test cases passed.")
test_experiment6()
```
