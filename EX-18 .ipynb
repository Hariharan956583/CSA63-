```python
 import socket
import time
import pandas as pd

print("="*60)
print("      SIMPLE VULNERABILITY ASSESSMENT")
print("            PORT SCANNER")
print("="*60)


target = input("\nEnter Target IP Address or Domain: ")

start_port = int(input("Enter Starting Port: "))
end_port = int(input("Enter Ending Port: "))

print("\nScanning Target:", target)
print("Port Range:", start_port, "-", end_port)


open_ports = []

start_time = time.time()

for port in range(start_port, end_port + 1):

    sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)

    sock.settimeout(0.5)

    result = sock.connect_ex((target, port))

    if result == 0:
        print(f"Port {port} : OPEN")
        open_ports.append(port)

    sock.close()

end_time = time.time()


services = {
    20:"FTP Data",
    21:"FTP",
    22:"SSH",
    23:"Telnet",
    25:"SMTP",
    53:"DNS",
    80:"HTTP",
    110:"POP3",
    123:"NTP",
    135:"RPC",
    139:"NetBIOS",
    143:"IMAP",
    161:"SNMP",
    389:"LDAP",
    443:"HTTPS",
    445:"SMB",
    3306:"MySQL",
    3389:"RDP",
    5432:"PostgreSQL",
    8080:"HTTP Alternate"
}


report = []

for port in open_ports:

    service = services.get(port, "Unknown")

    report.append([port, service, "Open"])

df = pd.DataFrame(report,
                  columns=["Port",
                           "Service",
                           "Status"])

print("\n")
print("="*60)
print("SCAN SUMMARY")
print("="*60)

if len(df) == 0:

    print("No Open Ports Found.")

else:

    print(df)

print("\nTotal Open Ports :", len(open_ports))

print("Time Taken : {:.2f} Seconds".format(end_time-start_time))


filename = "port_scan_report.csv"

df.to_csv(filename,index=False)

print("\nReport Saved As :", filename)

print("\nScan Completed Successfully.")
```
