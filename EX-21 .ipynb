```python
 import pandas as pd

print("="*75)
print(" HYPOTHESIS-, INTELLIGENCE-, AND ANALYTICS-BASED THREAT HUNTING")
print("="*75)
logs = pd.DataFrame({
    "Timestamp":[
        "10:00:01","10:00:05","10:00:08","10:00:10",
        "10:00:12","10:00:15","10:00:18","10:00:20",
        "10:00:25","10:00:30"
    ],
    "IP":[
        "192.168.1.10",
        "192.168.1.20",
        "192.168.1.10",
        "10.10.10.5",
        "192.168.1.10",
        "192.168.1.20",
        "10.10.10.5",
        "192.168.1.10",
        "172.16.1.15",
        "192.168.1.10"
    ],
    "Event":[
        "Failed Login",
        "Successful Login",
        "Failed Login",
        "Port Scan",
        "Failed Login",
        "File Access",
        "Malware Download",
        "Failed Login",
        "Normal Activity",
        "Failed Login"
    ]
})

print("\nSecurity Logs\n")
print(logs)

threat_feed = {
    "10.10.10.5":"Known Malware Server",
    "172.16.1.200":"Botnet Node",
    "185.199.110.153":"Command & Control Server"
}

print("\nThreat Intelligence Feed")
print(threat_feed)

failed = logs[logs["Event"]=="Failed Login"]

failed_count = failed.groupby("IP").size().reset_index(name="Failed_Logins")

results = []

for _, row in failed_count.iterrows():

    ip = row["IP"]
    failed_attempts = row["Failed_Logins"]

    intel = threat_feed.get(ip,"Not Found")

    risk = 20

    if failed_attempts >= 3:
        risk += 40

    if intel != "Not Found":
        risk += 40

    if risk >= 80:
        level = "HIGH"
    elif risk >= 50:
        level = "MEDIUM"
    else:
        level = "LOW"

    results.append([
        ip,
        failed_attempts,
        intel,
        risk,
        level
    ])

report = pd.DataFrame(results,
                      columns=[
                          "IP Address",
                          "Failed Logins",
                          "Threat Intelligence",
                          "Risk Score",
                          "Risk Level"
                      ])
print("\n")
print("="*75)
print("THREAT HUNTING REPORT")
print("="*75)

print(report)
print("\n")
print("="*75)
print("HIGH RISK ALERTS")
print("="*75)

alerts = report[report["Risk Level"]=="HIGH"]

if len(alerts)==0:
    print("No High Risk Threats Detected.")

else:
    for _, row in alerts.iterrows():
        print("IP Address           :", row["IP Address"])
        print("Failed Logins        :", row["Failed Logins"])
        print("Threat Intelligence  :", row["Threat Intelligence"])
        print("Risk Score           :", row["Risk Score"])
        print("Risk Level           :", row["Risk Level"])
        print("-"*50)

report.to_csv("Threat_Hunting_Report.csv", index=False)

print("\nReport Saved As : Threat_Hunting_Report.csv")

print("\nProgram Completed Successfully.")
```
