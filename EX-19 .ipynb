```python
import pandas as pd
import ipaddress
import re

print("="*70)
print(" MINI THREAT INTELLIGENCE PLATFORM")
print(" IOC NORMALIZATION & CORRELATION")
print("="*70)


ioc_data = [
    {"IOC":"192.168.1.100","Type":"IP"},
    {"IOC":"8.8.8.8","Type":"IP"},
    {"IOC":"HTTP://Malicious.COM/Login","Type":"URL"},
    {"IOC":"https://google.com","Type":"URL"},
    {"IOC":"ABCD1234EFGH5678IJKL9999AAAA1111","Type":"HASH"},
    {"IOC":"abcd1234efgh5678ijkl9999aaaa1111","Type":"HASH"},
    {"IOC":"evil@phishing.com","Type":"EMAIL"},
    {"IOC":"admin@gmail.com","Type":"EMAIL"}
]

df = pd.DataFrame(ioc_data)

print("\nOriginal IOC Data\n")
print(df)

def normalize(ioc, ioc_type):

    if ioc_type=="IP":
        try:
            return str(ipaddress.ip_address(ioc))
        except:
            return "Invalid"

    elif ioc_type=="URL":
        ioc=ioc.lower()
        ioc=ioc.replace("http://","")
        ioc=ioc.replace("https://","")
        ioc=ioc.rstrip("/")
        return ioc

    elif ioc_type=="HASH":
        return ioc.lower()

    elif ioc_type=="EMAIL":
        return ioc.lower()

    else:
        return ioc

df["Normalized_IOC"] = df.apply(
    lambda x: normalize(x["IOC"],x["Type"]),
    axis=1
)


threat_db = {
    "192.168.1.100":"Known Botnet",
    "malicious.com/login":"Phishing Website",
    "abcd1234efgh5678ijkl9999aaaa1111":"Trojan Malware",
    "evil@phishing.com":"Phishing Email"
}


status=[]
threat=[]
score=[]

for ioc in df["Normalized_IOC"]:

    if ioc in threat_db:

        status.append("Matched")
        threat.append(threat_db[ioc])
        score.append(90)

    else:

        status.append("Not Matched")
        threat.append("No Threat Found")
        score.append(10)

df["Correlation"]=status
df["Threat"]=threat
df["Risk Score"]=score
print("\n")
print("="*70)
print("NORMALIZED IOC RESULTS")
print("="*70)

print(df)

print("\n")
print("="*70)
print("SUMMARY")
print("="*70)

print("Total IOCs          :",len(df))
print("Threat Matches      :",len(df[df["Correlation"]=="Matched"]))
print("Safe Indicators     :",len(df[df["Correlation"]=="Not Matched"]))

filename="Threat_Intelligence_Report.csv"

df.to_csv(filename,index=False)

print("\nReport Saved As :",filename)

print("\nProgram Completed Successfully.")
```
