```python
 import pandas as pd
tools = [
["Vulnerability Assessment","Nmap","Network Scanner","Scans hosts and open ports"],
["Penetration Testing","Metasploit","Exploitation Framework","Tests system security"],
["Network Monitoring","Wireshark","Packet Analyzer","Captures network traffic"],
["Web Application Testing","Burp Suite","Web Security","Tests web vulnerabilities"],
["Password Auditing","John the Ripper","Password Cracker","Audits password strength"],
["Malware Analysis","VirusTotal","Malware Scanner","Analyzes suspicious files"],
["Log Analysis","Splunk","SIEM","Analyzes security logs"],
["Threat Detection","Snort","Intrusion Detection","Detects network attacks"],
["Digital Forensics","Autopsy","Forensics Tool","Investigates digital evidence"],
["Cloud Security","ScoutSuite","Cloud Auditor","Audits cloud environments"],
["Firewall Management","pfSense","Firewall","Network firewall management"],
["Incident Response","TheHive","Incident Platform","Manages security incidents"]
]

df = pd.DataFrame(
    tools,
    columns=["Task","Recommended Tool","Category","Purpose"]
)

df.to_csv("Security_Tools.csv", index=False)

print("Security Tool Database Created Successfully.\n")

print("Available IT Security Tasks\n")

for task in df["Task"]:
    print("-", task)

user_task = input("\nEnter Security Task : ")

result = df[df["Task"].str.lower() == user_task.lower()]

print("\n==============================")

if len(result) > 0:

    print("Recommended Security Tool\n")

    print(result.to_string(index=False))

else:

    print("No matching task found.")

print("==============================")

result.to_csv("Recommended_Tool.csv", index=False)

print("\nRecommendation saved as Recommended_Tool.csv")

print("\nTotal Available Tasks :", len(df))
print("Total Security Tools :", len(df["Recommended Tool"].unique()))

print("\nProgram Completed Successfully.")
```
