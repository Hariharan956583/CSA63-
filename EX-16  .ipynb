```python


import pandas as pd

print("=" * 60)
print("      DETECT POSSIBLE DoS ATTACK FROM LOG DATA")
print("=" * 60)


log_data = {
    "Timestamp": [
        "10:00:01","10:00:02","10:00:03","10:00:04","10:00:05",
        "10:00:06","10:00:07","10:00:08","10:00:09","10:00:10",
        "10:00:11","10:00:12","10:00:13","10:00:14","10:00:15"
    ],
    "IP_Address": [
        "192.168.1.10",
        "192.168.1.20",
        "192.168.1.10",
        "192.168.1.30",
        "192.168.1.10",
        "192.168.1.10",
        "192.168.1.20",
        "192.168.1.10",
        "192.168.1.10",
        "192.168.1.40",
        "192.168.1.10",
        "192.168.1.10",
        "192.168.1.20",
        "192.168.1.10",
        "192.168.1.10"
    ],
    "Request": [
        "GET","GET","POST","GET","GET",
        "POST","GET","GET","POST","GET",
        "GET","GET","POST","GET","GET"
    ]
}

df = pd.DataFrame(log_data)

print("\nServer Log Data\n")
print(df)


request_count = df["IP_Address"].value_counts().reset_index()
request_count.columns = ["IP_Address", "Total_Requests"]

print("\nRequest Count per IP Address\n")
print(request_count)


THRESHOLD = 5

print("\nDetection Threshold =", THRESHOLD)


suspicious = request_count[
    request_count["Total_Requests"] > THRESHOLD
]

print("\n" + "="*60)

if len(suspicious) == 0:
    print("No DoS Attack Detected.")

else:
    print("Possible DoS Attack Detected!\n")

    for index, row in suspicious.iterrows():
        print("IP Address      :", row["IP_Address"])
        print("Total Requests  :", row["Total_Requests"])
        print("Status          : Suspicious")
        print("-"*40)

print("="*60)

request_count.to_csv("dos_detection_report.csv", index=False)

print("\nReport saved as : dos_detection_report.csv")

print("\nProgram Completed Successfully.")
```
