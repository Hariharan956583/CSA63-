```python
 IDS_RULES = [
    {
        "sid": 2000001,
        "msg": "Possible SQL Injection",
        "proto": "tcp",
        "dst_port": 80,
        "content": "union select"
    },
    {
        "sid": 2000002,
        "msg": "Directory Traversal Attempt",
        "proto": "tcp",
        "dst_port": 80,
        "content": "../../../etc/passwd"
    },
    {
        "sid": 2000003,
        "msg": "Suspicious RDP Brute Force Pattern",
        "proto": "tcp",
        "dst_port": 3389,
        "content": "login_attempt"
    }
]


def scan_packet(packet, rules=IDS_RULES):
    """
    packet = {
        "proto": protocol,
        "dst_port": destination_port,
        "payload": packet_data
    }

    A rule triggers only if:
    1. Protocol matches
    2. Destination port matches
    3. Content pattern exists in payload
    """

    alerts = []

    payload_lower = packet["payload"].lower()

    for rule in rules:

        if (
            packet["proto"] == rule["proto"]
            and packet["dst_port"] == rule["dst_port"]
        ):

            if rule["content"].lower() in payload_lower:
                alerts.append(rule["msg"])

    return alerts
 def test_experiment3():
    sqli_packet = {
        "proto": "tcp",
        "dst_port": 80,
        "payload": "id=1' UNION SELECT user,pass FROM accounts--"
    }

    result1 = scan_packet(sqli_packet)
    print("SQL Injection Test:", result1)

    assert "Possible SQL Injection" in result1
    traversal_packet = {
        "proto": "tcp",
        "dst_port": 80,
        "payload": "GET /files?path=../../../etc/passwd"
    }
    result2 = scan_packet(traversal_packet)
    print("Directory Traversal Test:", result2)

    assert "Directory Traversal Attempt" in result2
    wrong_context = {
        "proto": "tcp",
        "dst_port": 22,
        "payload": "id=1' UNION SELECT user,pass FROM accounts--"
    }

    result3 = scan_packet(wrong_context)
    print("Wrong Context Test:", result3)

    assert result3 == []

    print("\nAll test cases passed.")
test_experiment3()
```
