```python
import hashlib
import os

def calculate_file_hash(filepath, algorithm='sha256'):
    """
    Calculates the cryptographic hash of a file.

    Args:
        filepath (str): The path to the file.
        algorithm (str): The hashing algorithm to use (e.g., 'md5', 'sha1', 'sha256').

    Returns:
        str: The hexadecimal representation of the file's hash.
             Returns None if the file does not exist.
    """
    if not os.path.exists(filepath):
        print(f"Error: File not found at {filepath}")
        return None

    hash_func = getattr(hashlib, algorithm)()
    with open(filepath, 'rb') as f:
        for chunk in iter(lambda: f.read(4096), b''):
            hash_func.update(chunk)
    return hash_func.hexdigest()

print("Function 'calculate_file_hash' defined.")
```

```python
file_name = 'my_important_document.txt'

# 1. Create a sample file and calculate initial hash
original_content = "This is a very important document that must not be altered."
with open(file_name, 'w') as f:
    f.write(original_content)

print(f"Original file '{file_name}' created.")
initial_hash = calculate_file_hash(file_name)
print(f"Initial SHA256 Hash: {initial_hash}\n")

# 2. Simulate file tampering and calculate new hash
tampered_content = "This is a very important document that has been altered!"
with open(file_name, 'w') as f:
    f.write(tampered_content)

print(f"File '{file_name}' has been tampered with.")
tampered_hash = calculate_file_hash(file_name)
print(f"Tampered SHA256 Hash: {tampered_hash}\n")

# 3. Compare hashes to detect tampering
if initial_hash == tampered_hash:
    print("Result: File integrity PASSED. The file has NOT been tampered with.")
else:
    print("Result: File integrity FAILED. The file HAS been tampered with!")

# Clean up the created file
os.remove(file_name)
print(f"\nCleaned up: Removed '{file_name}'.")
```
